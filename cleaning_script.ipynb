Load and explore the data

```python
import pandas as pd

df = pd.read_csv('Sample - Superstore.csv.zip', compression='zip', encoding='latin1')
# First look
print(df.shape)       # how many rows and columns
print(df.head())      # first 5 rows
print(df.info())      # column types and null counts
print(df.describe())  # basic statistics
```

Check for and fix issues

```python
# Check for missing values
print(df.isnull().sum())
```

```python
# Check for duplicate rows
print(df.duplicated().sum())
```

```python
# Remove duplicates if any
df = df.drop_duplicates()
```

```python
# Fix date columns (convert text to proper date format)
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
# Add a new column: profit margin percentage
df['Profit Margin %'] = (df['Profit'] / df['Sales'] * 100).round(2)
```

```python
# Add year and month columns (useful for Power BI later)
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month_name()
```

Save the cleaned file

```python
df.to_csv('superstore_cleaned.csv', index=False)
print("Cleaned file saved! Shape:", df.shape)

# Download it to your computer
from google.colab import files
files.download('superstore_cleaned.csv')
```
